## 6. 📄 Papers Fundamentales

### Paper 1: Reinforcement Learning: An Introduction

**Sutton & Barto (2018)**  
[Libro completo gratis](http://incompleteideas.net/book/)

**Contribución:**
- LA referencia definitiva en RL
- Capítulos 1-2 cubren fundamentos de agentes
- Explicaciones claras con ejemplos

**Impacto:** Libro de texto estándar en cursos de RL mundialmente.

---

### Paper 2: Playing Atari with Deep RL

**Mnih et al., DeepMind (2013)**  
[arXiv:1312.5602](https://arxiv.org/abs/1312.5602)

**Contribución:**
- Primer agente que aprende a jugar Atari desde píxeles
- Combina Deep Learning + RL
- Arquitectura DQN

**Impacto:** Inicio de la era Deep RL moderna.


## 7. ✅ Best Practices

### 1. Diseño de Ambientes

**✅ DO:**
- Interfaz clara: reset(), step(), render()
- Estados/acciones bien definidos
- Recompensas informativas
- Terminación clara

**❌ DON'T:**
- Recompensas ambiguas
- Estados ilimitados sin terminación

---

### 2. Políticas

**Random:** Baseline, buena para exploración
**Greedy:** Simple, puede atascarse
**Learning:** Mejora con experiencia

---

### 3. Debugging RL

```python
# Verificar ambiente
assert env.reset() in env.state_space

# Verificar episodio completo
state = env.reset()
for _ in range(100):
    action = agent.select_action(state)
    state, reward, done, _ = env.step(action)
    if done: break
```

3. **Ciclo:** Observar → Decidir → Actuar → Repetir
4. **Política:** Estrategia del agente
