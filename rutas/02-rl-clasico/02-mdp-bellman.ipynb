**Contribución:** Introduce MDPs y programación dinámica.

**Impacto:** Base matemática del RL.

### Howard (1960) - Dynamic Programming

## 7. ✅ Best Practices

### Diseño de MDPs
**✅ DO:**
- Estados Markovianos
- Recompensas informativas
- γ ∈ [0.9, 0.99] típicamente

**❌ DON'T:**
- Estados que violan Markov
